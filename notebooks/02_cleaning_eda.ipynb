## Imbalance, bias, and limitations

- CFPB submissions do not represent all customers or product users.
- Product and Issue distributions are strongly imbalanced and taxonomies change over time.
- Reporting patterns and dataset coverage can change, so time trends mix behavior and coverage effects.
- Day 5 rejected 13,210,233 rows without usable narratives; retained-text EDA therefore has material selection bias.
- Complaint counts alone do not prove actual service quality or population rates.
- The seven cleaned columns do not include company, geography, submission channel, response, or denominator data.
- Department-label imbalance is outside this Day 6 analysis; Day 7 subsequently quantified the deterministic proxy-label distribution in `docs/label_mapping.md`.
- Narrative percentiles use the deterministic lower nearest-rank value at one-based rank `floor((n - 1) * p) + 1`, without interpolation.
- Day 6 implementation and review are complete. Day 7 subsequently froze deterministic mapping and dataset version `v1`; Day 8 model training remains unstarted.
